```python
def sp(two_lines = True):
    print("==========================")
    if (two_lines):
        print()
        print("==========================")
```

## Data

```python
def get_info(df: pd.DataFrame):
    print("Shape:", df.shape)
    print("Columns:", df.columns)
    sp()
    print(df.info())
    sp()
    print(df.head())
    sp()
    print(df.describe())
    sp()
    print(df.isnull().sum())
    sp()
    print("Total Null quantity:", df.isnull().sum().sum())
```

```python
df_orig, df_test_orig = _data.get_data('orig')

get_info(df_orig)
```

```python
print(f"Fraud rate: {round(df_orig['is_fraud'].mean(), 4) * 100}%")
```

### Editing string values

```python
def processing_dataset(df_orig: pd.DataFrame):
    df = df_orig.copy()
    df = df.drop(columns='id')
    for col in df_orig.dtypes[df_orig.dtypes == 'str'].index:
        df[col] = df[col].str.lower().str.replace(' ', '_')
    return df
```

```python
df = processing_dataset(df_orig)
df_test = processing_dataset(df_test_orig)
```
